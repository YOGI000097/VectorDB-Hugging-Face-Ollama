Q1: What is a Vector Database (VectorDB)?

A Vector Database stores data as embeddings (vectors) instead of rows/columns like traditional databases.

| Feature    | Traditional DB    | Vector DB                |
| ---------- | ----------------- | ------------------------ |
| Data Type  | Structured        | High-dimensional vectors |
| Query Type | Exact match (SQL) | Similarity search        |
| Use Case   | Banking, ERP      | AI, NLP, search engines  |


Q2: Types of VectorDBs
1. Approximate Nearest Neighbor (ANN)
Fast search, slightly less accurate
Example: FAISS
Use case: Large-scale AI systems
2. Exact Search
100% accurate but slower
Use case: small datasets
3. Cloud-based Vector DB
Scalable, managed
Example: Pinecone
4. Embedded DB
Runs locally
Example: ChromaDB

Q5: Using Hugging Face Models

Easy fine-tuning

```python
# Q6: ChromaDB Setup + Semantic Search
# 🔹 Install:
!pip install chromadb sentence-transformers

import chromadb
from sentence_transformers import SentenceTransformer

# Initialize model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Create client
client = chromadb.Client()

collection = client.create_collection("demo")

# Sample data
docs = ["AI is amazing", "Machine learning is powerful", "I love programming"]
embeddings = model.encode(docs)

# Insert into DB
for i, doc in enumerate(docs):
    collection.add(
        documents=[doc],
        embeddings=[embeddings[i]],
        ids=[str(i)]
    )

# Query
query = "AI technology"
query_embedding = model.encode([query])

results = collection.query(query_embeddings=query_embedding, n_results=2)

print(results)
```

```python
# Q7: Hugging Face Fine-Tuning (Text Generation)
# # 🔹 Install:
pip install transformers datasets torch
from transformers import AutoModelForCausalLM, AutoTokenizer, Trainer, TrainingArguments
from datasets import load_dataset

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

dataset = load_dataset("wikitext", "wikitext-2-raw-v1", split="train[:1%]")

def tokenize(example):
    return tokenizer(example["text"], truncation=True, padding="max_length")

tokenized_data = dataset.map(tokenize, batched=True)

training_args = TrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=2,
    num_train_epochs=1
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_data
)

trainer.train()
```

```python
# Q9: Basic RAG using Ollama + Chroma

import chromadb
import requests
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
client = chromadb.Client()
collection = client.create_collection("rag")

docs = ["AI is future", "Geoinformatics helps disaster management"]
embeddings = model.encode(docs)

for i, doc in enumerate(docs):
    collection.add(documents=[doc], embeddings=[embeddings[i]], ids=[str(i)])

# Query
query = "What is AI?"
query_embedding = model.encode([query])

results = collection.query(query_embeddings=query_embedding, n_results=1)

context = results['documents'][0][0]

# Send to Ollama
response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3",
        "prompt": f"Context: {context}\nQuestion: {query}\nAnswer:"
    }
)

print(response.json()["response"])
```

Q10: Health-Tech Chatbot Solution

🔹 Architecture:
Hugging Face
Convert medical articles → embeddings
VectorDB (Chroma)
Store and retrieve relevant documents
Ollama
Generate answers using LLM

```python
from sentence_transformers import SentenceTransformer
import chromadb
import requests

# Step 1: Embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')

docs = ["COVID-19 affects lungs", "Diabetes increases sugar levels"]
embeddings = model.encode(docs)

# Step 2: Store in VectorDB
client = chromadb.Client()
collection = client.create_collection("medical")

for i, doc in enumerate(docs):
    collection.add(documents=[doc], embeddings=[embeddings[i]], ids=[str(i)])

# Step 3: Query
query = "What is diabetes?"
query_embedding = model.encode([query])

results = collection.query(query_embeddings=query_embedding, n_results=1)
context = results['documents'][0][0]

# Step 4: Generate Answer
response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3",
        "prompt": f"Context: {context}\nQuestion: {query}\nAnswer:"
    }
)

print(response.json()["response"])
```
